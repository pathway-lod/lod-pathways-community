# Minimum information tiers: IDSM chemical similarity

Three queries that use the **IDSM / Sachem chemical structure search** (<https://idsm.elixir-czech.cz>) on ChEBI to link metabolites by structural similarity to the metabolites of PlantMetWiki pathways. Cases 5 (metabolomics), 6 and 4 (the last query, with a gene) and 2 (pathway components). Unlike the other notebooks, these queries run **directly on the public PlantMetWiki endpoint**, which calls IDSM with `SERVICE` (no evidence graph is loaded); the detected metabolite and the gene are given in the query as examples.

The outputs are a dated snapshot (see the summary at the end).

```python
import mi
import pandas as pd
from IPython.display import display
CASE = '5 metabolomics (IDSM)'
print('PlantMetWiki release:', mi.pmw_release())
```

## Similar metabolites of the conversions of a pathway

`idsm-similar-metabolites-in-pathway`

idsm-similar-metabolites-in-pathway: For the directed conversions of one PlantMetWiki pathway (here PC968, alpha-tomatine biosynthesis; change the identifier), finds the ChEBI compounds that are structurally similar (Sachem similarity of at least 0.98) to the source and to the target metabolite of each conversion, using the IDSM chemical structure search service. May be slower because it depends on the external IDSM endpoint. NOTE: this version looks at the substrates (wp:source) of the conversions; for the products change wp:source to wp:target and the role to product. When tested on 2026-11-11 the substrate version returned 19 rows in a few seconds, and the product version ended with a proxy error (502) of the PlantMetWiki endpoint, so it is not shipped as a separate query. OUTPUT: a list of metabolites (putative members of the pathway: ChEBI compounds structurally similar to a substrate or product of a conversion of the pathway), with the pathway, the pathway metabolite they are similar to, its role (substrate or product), the conversion and the similarity score as metadata.

*Keywords:* ChEBI, IDSM, Minimum Information: evidence tiers, PlantMetWiki pathway, Sachem, case 2: inferred enzyme function from PMN/PlantCyc+ db, case 5: metabolomics, metabolite annotations, chemical similarity search, conversion, evidence tier, federated query, metabolite, output: metabolites, runs on public endpoints, structure similarity, substrate and product

```python
r = mi.run_endpoint('idsm-similar-metabolites-in-pathway', CASE, timeout=150)
mi.record(r)
print(r['status'], '|', r['n_rows'], 'rows |', r['seconds'], 's')
display(r['table'].head(10))
```

## A detected metabolite is similar to a metabolite of a pathway

`metabolite-similar-to-pathway-metabolite-idsm`

metabolite-similar-to-pathway-metabolite-idsm: A detected metabolite is kept if the IDSM chemical structure search finds a ChEBI compound similar to it (Sachem similarity of at least 0.9) that is a metabolite of a PlantMetWiki pathway. Returns the similar compound, the score and the pathways. The detected metabolite is given by its SMILES in the two BIND lines (here 2,5-dihydroxybenzoate, a metabolite of the Brassicaceae annotation pilot, Wikidata Q27125148); in a run over the evidence graph the SMILES come from the annotations (for example from the Wikidata property P233). One metabolite per run, because the PlantMetWiki endpoint does not accept two SERVICE clauses with a VALUES block. OUTPUT: a list of metabolites (putative: the detected metabolite is kept as a putative member of a pathway when it is structurally similar to a metabolite of that pathway), with the pathway, the similar ChEBI compound and the similarity score as metadata.

*Keywords:* ChEBI, IDSM, Minimum Information: evidence tiers, PlantMetWiki metabolite, PlantMetWiki pathway, SMILES, Sachem, Wikidata, case 5: metabolomics, metabolite annotations, chemical similarity search, evidence tier, federated query, metabolite, metabolite annotation, output: metabolites, runs on public endpoints, structure similarity

```python
r = mi.run_endpoint('metabolite-similar-to-pathway-metabolite-idsm', CASE, timeout=150)
mi.record(r)
print(r['status'], '|', r['n_rows'], 'rows |', r['seconds'], 's')
display(r['table'].head(10))
```

## Similar metabolite, and a co-expressed gene that catalyses a conversion there

`idsm-similar-metabolite-with-coexpressed-catalysing-gene`

idsm-similar-metabolite-with-coexpressed-catalysing-gene: A detected metabolite is similar (IDSM, Sachem similarity of at least 0.7) to a metabolite of a PlantMetWiki pathway, and a gene that is co-expressed with it (associated with it in a functional cluster) encodes an enzyme (UniProt) that catalyses a conversion of that pathway in which the similar metabolite is a substrate or a product. The detected metabolite and the gene are given as one example in the three BIND lines (solasodine and the tomatidine 3-O-galactosyltransferase E5L3R8, both of alpha-tomatine biosynthesis); the co-expression association itself is not queried here, because the associations of the evidence graph have no annotation or UniProt link yet. OUTPUT: a list of reactions (putative or supported: a conversion of a pathway that involves a metabolite similar to the detected metabolite and is catalysed by an enzyme that a co-expressed gene encodes), with the pathway, the detected metabolite, the similar ChEBI compound, its score and the enzyme as metadata.

*Keywords:* ChEBI, IDSM, Minimum Information: evidence tiers, PlantMetWiki pathway, PlantMetWiki reaction, Sachem, UniProt, case 2: inferred enzyme function from PMN/PlantCyc+ db, case 4: transcriptomics, gene co-expression values, case 5: metabolomics, metabolite annotations, case 6: paired omics, transcript-metabolite associations, chemical similarity search, co-expression, enzyme, evidence tier, federated query, gene, gene-metabolite association, metabolite, output: reactions, runs on public endpoints, structure similarity, substrate and product

```python
r = mi.run_endpoint('idsm-similar-metabolite-with-coexpressed-catalysing-gene', CASE, timeout=150)
mi.record(r)
print(r['status'], '|', r['n_rows'], 'rows |', r['seconds'], 's')
display(r['table'].head(10))
```
